# Setup

```python
import numpy as np
import matplotlib.pyplot as plt

N, T = 3, 120                      # 3 restaurants, 120 days
true_rates = np.array([0.3, 0.5, 0.8])  # hidden satisfaction rates
np.random.seed(42)                   # reproducibility
```

#  Epsilon-Greedy (ε = 0.1)

```python
 # ε-Greedy — explore randomly with prob ε, exploit best known otherwise
counts_e, rewards_e, eps = np.zeros(N), np.zeros(N), 0.1
for _ in range(T):
    arm = np.random.randint(N) if np.random.rand() < eps else np.argmax(rewards_e / np.maximum(counts_e, 1))
    r = np.random.binomial(1, true_rates[arm]); counts_e[arm] += 1; rewards_e[arm] += r
```

# Upper Confidence Bound (UCB)

```python
# UCB — pick restaurant with highest mean + exploration bonus
counts_u, rewards_u = np.zeros(N), np.zeros(N)
for t in range(1, T+1):
    arm = np.argmax(rewards_u / np.maximum(counts_u,1) + np.sqrt(2*np.log(t) / np.maximum(counts_u,1)))
    r = np.random.binomial(1, true_rates[arm]); counts_u[arm] += 1; rewards_u[arm] += r
```

# Thompson Sampling

```python
# Thompson Sampling — sample Beta(α, β) per arm, pick highest sample
alpha, beta_p, rewards_t = np.ones(N), np.ones(N), 0
for _ in range(T):
    arm = np.argmax(np.random.beta(alpha, beta_p))
    r = np.random.binomial(1, true_rates[arm]); alpha[arm] += r; beta_p[arm] += 1-r; rewards_t += r
```
