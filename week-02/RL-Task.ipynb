# Setup

```python
import numpy as np
import matplotlib.pyplot as plt

N, T = 3, 120                      # 3 restaurants, 120 days
true_rates = np.array([0.3, 0.5, 0.8])  # hidden satisfaction rates
np.random.seed(42)                   # reproducibility
```

#  Epsilon-Greedy (ε = 0.1)

```python
 # ε-Greedy — explore randomly with prob ε, exploit best known otherwise
counts_e, rewards_e, eps = np.zeros(N), np.zeros(N), 0.1
for _ in range(T):
    arm = np.random.randint(N) if np.random.rand() < eps else np.argmax(rewards_e / np.maximum(counts_e, 1))
    r = np.random.binomial(1, true_rates[arm]); counts_e[arm] += 1; rewards_e[arm] += r
```
