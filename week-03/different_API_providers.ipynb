```python
from google import genai

from google.colab import userdata
gemini_key = userdata.get('GEMINI_API_KEY')

# The client gets the API key from the environment variable `GEMINI_API_KEY`.
client = genai.Client(api_key=gemini_key)

response = client.models.generate_content(
    model="gemini-3-flash-preview", contents="What is RLHF in training LLMs?"
)
print(response.text)
```

```python
!pip install groq
```

```python
import os
from groq import Groq
from google.colab import userdata

# Retrieve the secret securely from Colab's secret manager
# (Make sure the names match exactly, including capitalization)
groq_key = userdata.get('GROQ_API_KEY')

from groq import Groq

client = Groq(api_key=groq_key)
completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
      {
        "role": "user",
        "content": "What is RLHF in training LLMs?"
      }
    ],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

# Groq (OpenAI Compatible)

```python

from openai import OpenAI
import os

groq_key = userdata.get('GROQ_API_KEY')

client = OpenAI(
    api_key=groq_key,
    base_url="https://api.groq.com/openai/v1",
)

response = client.responses.create(
    input="Explain the importance of fast language models",
    model="openai/gpt-oss-20b",
)
print(response.output_text)
```

# Gemini (OpenAI Compatible)

```python
from openai import OpenAI

client = OpenAI(
    api_key=gemini_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

response = client.chat.completions.create(
    model="gemini-3-flash-preview",
    messages=[
        {   "role": "system",
            "content": "You are a helpful assistant."
        },
        {
            "role": "user",
            "content": "Explain to me how AI works"
        }
    ]
)

print(response.choices[0].message)
```

# OpenRouter API

```python
import requests
import json

open_router_key = userdata.get('OPEN_ROUTER_KEY')

response = requests.post(
  url="https://openrouter.ai/api/v1/chat/completions",
  headers={
    # 1. Added 'Bearer ' before the key
    "Authorization": f"Bearer {open_router_key}",
    # 2. Added content type so the server knows you are sending JSON
    "Content-Type": "application/json"
  },
  data=json.dumps({
    "model": "openai/gpt-oss-20b:free",
    "messages": [
      {
        "role": "user",
        "content": "What is the meaning of life?"
      }
    ]
  })
)

# View the output safely
if response.status_code == 200:
    response_json = response.json()
    reply = response_json['choices'][0]['message']['content']
    print(reply)
else:
    print(f"Error {response.status_code}: {response.text}")
```
