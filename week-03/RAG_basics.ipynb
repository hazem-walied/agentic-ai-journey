# Webpage content Loading (Scraping)



> The following methods are simple and not very efficient, we should do more work to make high quality scraping



```python
!pip install -q langchain-community beautifulsoup4 html2text
```

```python
from langchain_community.document_loaders import AsyncHtmlLoader
from langchain_community.document_transformers import BeautifulSoupTransformer

# 1. Define the URL(s) you want to scrape
urls = ["https://en.wikipedia.org/wiki/Artificial_intelligence"]

# 2. Fetch the raw HTML using AsyncHtmlLoader
print("Fetching HTML content...")
loader = AsyncHtmlLoader(urls)
docs = loader.load()

# 3. Initialize the BeautifulSoupTransformer to clean the content
bs_transformer = BeautifulSoupTransformer()

# 4. Transform the documents (removes tags, scripts, and extra whitespace)
print("Parsing and cleaning content with BeautifulSoup...")
docs_transformed = bs_transformer.transform_documents(
    docs,
    tags_to_extract=["p", "h1", "h2", "h3"]  # Focus only on headers and body paragraphs
)

# 5. View the output
print("\n--- Successfully Loaded Webpage ---")
print(f"Source URL: {docs_transformed[0].metadata['source']}")
print("\n--- Sample Content (First 500 characters) ---")
print(docs_transformed[0].page_content[:500])
```

## Another method also

```python
import os
from langchain_community.document_loaders import WebBaseLoader

# 1. Set a standard User-Agent so websites don't block the request
os.environ["USER_AGENT"] = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"

# 2. Initialize WebBaseLoader (it uses BeautifulSoup automatically under the hood)
url = "https://en.wikipedia.org/wiki/Artificial_intelligence"
loader = WebBaseLoader(url)

print("Fetching and parsing webpage...")
docs = loader.load()

# 3. View the output
print("\n--- Successfully Loaded Webpage ---")
print(f"Source URL: {docs[0].metadata['source']}")
print(f"Total Character Length: {len(docs[0].page_content)}")
print("\n--- Sample Content (First 500 characters) ---")

# Strip leading whitespaces just in case there's an empty header gap
content = docs[0].page_content.strip()
print(content[:1500])
```

# Another one (The most efficient for wikipidia)

```python
!pip install -q langchain-community wikipedia
```

```python
from langchain_community.document_loaders import WikipediaLoader

# 1. Initialize the loader with your search query
# We set load_max_docs=1 so it only grabs the most relevant matching page
loader = WikipediaLoader(query="Artificial intelligence", load_max_docs=1)

print("Searching and loading from Wikipedia...")
docs = loader.load()

# 2. Check the output
if docs:
    print("\n--- Successfully Loaded from Wikipedia ---")
    print(f"Title: {docs[0].metadata['title']}")
    print(f"Source URL: {docs[0].metadata['source']}")
    print(f"Total Character Length: {len(docs[0].page_content)}")
    print("\n--- Sample Content (First 500 characters) ---")
    print(docs[0].page_content[:500])
else:
    print("No matching Wikipedia pages found.")
```

# Text Splitters

```python
!pip install -U langchain-text-splitters pypdf
```

```python
from langchain_community.document_loaders import PyPDFLoader

# 1. Initialize the loader with your file path
loader = PyPDFLoader("attention.pdf")

print("Parsing PDF file...")
document = loader.load()

# 2. Verify the structure
print("\n--- Successfully Loaded PDF ---")
print(f"Total pages extracted: {len(document)}")

# Inspect the first page
print("\n--- First Page Content (First 500 characters) ---")
print(document[0].page_content[:500])

print("\n--- First Page Metadata ---")
print(document[0].metadata)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

# FIX: Use split_documents instead of split_text
texts = text_splitter.split_documents(document)

# Verify the result
print(f"Created {len(texts)} chunks.")
print("\n--- First Chunk Content ---")
print(texts[0].page_content)
print("\n--- First Chunk Metadata ---")
print(texts[0].metadata)  # Notice your page numbers/sources are preserved!
```
